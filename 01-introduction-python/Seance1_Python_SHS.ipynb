Avant de manipuler notre premier jeu de données, prenons le temps de nous familiariser avec quelques briques fondamentales du langage Python. Ces notions vous seront utiles tout au long de la journée — et bien au-delà, dès que vous voudrez adapter du code à vos propres besoins.

## 1.1 Variables et types de base

Une **variable** est un nom que l'on associe à une valeur, à l'aide du signe `=` (l'opérateur d'affectation). On peut ensuite réutiliser ce nom pour accéder à la valeur stockée.

````python
nom = "Marie"
age = 32
taille = 1.68
est_chercheuse = True

print(nom)
print(age)
````

Python distingue plusieurs **types** de base, que l'on peut toujours vérifier avec la fonction `type()` :

| Type | Description | Exemple |
|---|---|---|
| `str` | Chaîne de caractères (texte), entre guillemets simples ou doubles | `"Marie"` |
| `int` | Nombre entier | `32` |
| `float` | Nombre à virgule flottante (décimal) | `1.68` |
| `bool` | Valeur booléenne (vrai/faux) | `True` / `False` |


````python
type(nom)
type(age)
type(taille)
type(est_chercheuse)
````

On peut effectuer des opérations arithmétiques de base (`+`, `-`, `*`, `/`) et construire des phrases combinant texte et variables grâce aux **f-strings** (chaînes de caractères "formatées"), en faisant précéder les guillemets de la lettre `f` et en plaçant les variables entre accolades `{}` :

````python
annee_actuelle = 2026
annee_naissance = annee_actuelle - age

message = f"{nom} a {age} ans et est née en {annee_naissance}."
print(message)
````

## 1.2 Comparaisons et opérateurs logiques

Les **opérateurs de comparaison** (`==` égal à, `!=` différent de, `<`, `>`, `<=`, `>=`) permettent de comparer deux valeurs ; le résultat est toujours un booléen (`True` ou `False`) :

````python
print(age == 32)     # est égal à
print(age != 30)     # est différent de
print(age > 40)      # est supérieur à
````

<div class="alert alert-danger" role="alert" style="background-color:#f8d7da;padding:10px;border-radius:5px;">
⚠️ Ne confondez pas <code>=</code> (l'opérateur d'<b>affectation</b>, qui attribue une valeur à une variable) et <code>==</code> (l'opérateur de <b>comparaison</b>, qui teste une égalité) — une source d'erreur très fréquente, y compris chez les personnes expérimentées !
</div>

On peut combiner plusieurs conditions avec les opérateurs logiques `and` (ET), `or` (OU) et `not` (NON) :

````python
print(age > 18 and est_chercheuse)      # les deux conditions doivent être vraies
print(age > 65 or est_chercheuse)       # au moins une condition doit être vraie
print(not est_chercheuse)               # inverse la valeur booléenne
````

## 1.3 Les listes

Une **liste** est une collection **ordonnée** de valeurs, entre crochets `[ ]`, séparées par des virgules. Une liste peut contenir des éléments de n'importe quel type, y compris mélangés :

````python
universites = ["Harvard", "Yale", "Columbia", "MIT"]

print(universites)
print(type(universites))
print(len(universites))   # nombre d'éléments dans la liste
````

On accède à un élément d'une liste par sa **position** (son *index*), entre crochets. **Attention : en Python, le comptage commence à 0**, et non à 1 :

````python
print(universites[0])    # premier élément
print(universites[1])    # deuxième élément
print(universites[-1])   # dernier élément (les index négatifs comptent depuis la fin)
````

On peut aussi extraire une **tranche** (*slice*) de plusieurs éléments consécutifs, avec la notation `[début:fin]` (la borne de fin est exclue) :

````python
print(universites[0:2])   # les deux premiers éléments
print(universites[1:])   # à partir du deuxième élément jusqu'à la fin
````

Les listes sont **modifiables** (*mutables*) : on peut y ajouter, retirer ou modifier des éléments après leur création :

````python
universites.append("Princeton")   # ajoute un élément à la fin
print(universites)

universites[0] = "Stanford"       # remplace le premier élément
print(universites)

universites.remove("Yale")        # retire un élément précis
print(universites)
````

## 1.4 Les tuples

Un **tuple** ressemble beaucoup à une liste (une collection ordonnée d'éléments), mais s'écrit entre parenthèses `( )` et, à la différence de la liste, est **immuable** : une fois créé, on ne peut plus le modifier. On l'utilise typiquement pour représenter une collection de valeurs fixes, qui ne doivent pas changer (par exemple, des coordonnées géographiques) :

````python
coordonnees_paris = (48.8566, 2.3522)   # (latitude, longitude)

print(coordonnees_paris)
print(coordonnees_paris[0])   # accès par position, comme pour une liste

# coordonnees_paris[0] = 45.0   # décommenter cette ligne produirait une erreur : un tuple ne se modifie pas
````

## 1.5 Les dictionnaires

Un **dictionnaire** est une collection de paires **clé-valeur**, entre accolades `{ }`. Contrairement à une liste, on n'accède pas à un élément par sa position, mais par sa **clé** — un peu comme un vrai dictionnaire, où l'on cherche un mot (la clé) pour trouver sa définition (la valeur) :

````python
etudiant = {
    "nom": "Chen Wei",
    "nationalite": "Chinese",
    "annee_debut": 1928,
    "annee_fin": 1932
}

print(etudiant)
print(etudiant["nom"])            # accès à une valeur par sa clé
print(etudiant["annee_fin"] - etudiant["annee_debut"])   # les valeurs peuvent être utilisées dans des calculs
````

On peut ajouter ou modifier une entrée, lister les clés ou les valeurs, et parcourir l'ensemble des paires :

````python
etudiant["universite"] = "Columbia"   # ajoute une nouvelle paire clé-valeur
print(etudiant)

print(etudiant.keys())     # liste des clés
print(etudiant.values())   # liste des valeurs

for cle, valeur in etudiant.items():   # parcourir toutes les paires clé-valeur
    print(cle, "->", valeur)
````

## 1.6 Les conditions : `if` / `elif` / `else`

Une **structure conditionnelle** permet d'exécuter un bloc de code seulement si une condition est vraie. En Python, l'indentation (les espaces en début de ligne) n'est pas une question de style : elle indique **quelles instructions appartiennent à quel bloc**. C'est une particularité importante du langage à bien garder en tête.

````python
annee_debut = 1936

if annee_debut < 1914:
    print("Avant la Première Guerre mondiale")
elif annee_debut <= 1918:
    print("Pendant la Première Guerre mondiale")
else:
    print("Après la Première Guerre mondiale")
````

<div class="alert alert-success" role="alert" style="background-color:#d4edda;padding:10px;border-radius:5px;">
💡 <code>elif</code> (contraction de "else if") permet d'enchaîner plusieurs conditions ; <code>else</code> couvre tous les cas restants. On peut n'avoir qu'un simple <code>if</code>, ou combiner autant de <code>elif</code> que nécessaire, mais un seul <code>else</code> final.
</div>

## 1.7 Les boucles : `for` et `while`

Une **boucle `for`** permet de répéter une instruction pour chaque élément d'une collection (liste, dictionnaire...) :

````python
universites = ["Harvard", "Columbia", "MIT"]

for u in universites:
    print(f"Université : {u}")
````

La fonction `range()` génère une séquence de nombres, très utile pour répéter une action un nombre défini de fois :

````python
for i in range(5):    # de 0 à 4 (5 valeurs, borne finale exclue)
    print(i)
````

On peut aussi accumuler un résultat au fil d'une boucle — par exemple, calculer la somme des durées de plusieurs formations :

````python
durees = [4, 2, 3, 5, 2]
total = 0

for d in durees:
    total = total + d

print(f"Durée totale : {total} ans")
````

Une **boucle `while`** répète un bloc d'instructions **tant qu'**une condition reste vraie :

````python
compteur = 0

while compteur < 5:
    print(f"Itération {compteur}")
    compteur = compteur + 1   # ne pas oublier d'incrémenter, sous peine de boucle infinie !
````

<div class="alert alert-danger" role="alert" style="background-color:#f8d7da;padding:10px;border-radius:5px;">
⚠️ Avec une boucle <code>while</code>, veillez toujours à ce que la condition finisse par devenir fausse — sans quoi la boucle tournera indéfiniment (une <b>boucle infinie</b>), obligeant à interrompre manuellement l'exécution (bouton "Interrompre" ou <code>Kernel &gt; Interrupt</code> dans Jupyter).
</div>

## 1.8 Les fonctions

Une **fonction** est un bloc de code réutilisable, que l'on définit une fois avec le mot-clé `def`, puis que l'on peut appeler autant de fois que nécessaire — plutôt que de recopier le même code à chaque fois. Une fonction peut prendre des **arguments** (des valeurs en entrée) et **renvoyer** un résultat avec `return` :

````python
def calculer_duree(annee_debut, annee_fin):
    duree = annee_fin - annee_debut
    return duree

# On peut maintenant appeler cette fonction autant de fois que nécessaire
print(calculer_duree(1928, 1932))
print(calculer_duree(1910, 1914))
````

On peut donner une **valeur par défaut** à un argument, utilisée si l'on n'en fournit pas explicitement lors de l'appel :

````python
def saluer(nom, langue="français"):
    if langue == "français":
        return f"Bonjour, {nom} !"
    elif langue == "anglais":
        return f"Hello, {nom}!"
    else:
        return f"({langue} non pris en charge)"

print(saluer("Chen Wei"))
print(saluer("Chen Wei", langue="anglais"))
````

<div class="alert alert-success" role="alert" style="background-color:#d4edda;padding:10px;border-radius:5px;">
💡 Écrire ses propres fonctions permet d'éviter de dupliquer du code, de le rendre plus lisible, et de le réutiliser facilement d'un projet à l'autre. Nous en écrirons plusieurs au fil de la journée — et davantage encore lors des prochaines séances (fouille de texte, réseaux...).

La méthode **`.assign()`** (ou l'affectation directe `df["nouvelle_col"] = ...`) est couramment employée pour créer de nouvelles colonnes — c'est l'équivalent de `mutate()`.

## Comptages groupés

Avec pandas, il est très facile de faire des comptages complexes croisant plusieurs variables, en combinant **`.groupby()`** et **`.size()`** (ou en appliquant directement `.value_counts()`). Par exemple, examinons la distribution des étudiants par discipline :

````python
auc_simple["Field"].value_counts()    # trié par défaut par ordre décroissant

# Équivalent avec groupby :
# auc_simple.groupby("Field").size().sort_values(ascending=False)
````

La combinaison **`.groupby()` + `.size()`/`.agg()`** est l'équivalent des **tableaux croisés dynamiques** obtenus avec un logiciel comme Excel, mais offre beaucoup plus de latitude dans le choix des variables à combiner.

Si le résultat contient beaucoup de valeurs, on peut le stocker dans une nouvelle variable :

````python
field_count = auc_simple["Field"].value_counts().reset_index() 
field_count
````

On peut également calculer le pourcentage que représente chaque discipline et le stocker dans une colonne :

````python
field_count["percent"] = field_count["count"] / field_count["count"].sum() * 100
field_count
````

On peut enfin utiliser **`.groupby()`** pour créer de nouvelles variables synthétiques, par exemple calculer la durée totale d'études de chaque étudiant. C'est ici que **`.agg()`** entre en jeu (équivalent de `summarise()`) :

````python
summary_years = auc_simple.groupby(["Name", "Nationality"]).agg(
    study_start_year=("Start", "min"),
    study_end_year=("End", "max")
).reset_index()

summary_years["study_duration"] = summary_years["study_end_year"] - summary_years["study_start_year"]

summary_years
````

<div class="alert alert-success" role="alert" style="background-color:#d4edda;padding:10px;border-radius:5px;">
<b>Explication</b>

- <code>.groupby(["Name", "Nationality"])</code> : on regroupe les lignes du tableau par combinaison de Name et Nationality. Autrement dit, toutes les observations relatives à un même étudiant seront traitées ensemble.
- <code>.agg(nouvelle_col=("colonne_source", "fonction"))</code> sert à réduire un tableau de données en un résumé, en appliquant une fonction d'agrégation (min, max, mean, sum...) sur chaque groupe. Après <code>.agg()</code>, on obtient une seule ligne par groupe.
- <code>.reset_index()</code> transforme les colonnes de regroupement (qui deviennent temporairement un "index") en colonnes normales du dataframe — un réflexe fréquent après un <code>.groupby()</code> en pandas.
- Par défaut, pandas <b>ignore les valeurs manquantes</b> (NA) dans les fonctions d'agrégation comme min/max/mean.
</div>

## Compter avec les fonctions de base

Pandas reste très efficace pour des comptages simples sur les variables quantitatives. Par exemple, on peut facilement calculer la durée minimale, maximale et moyenne de la durée des formations :

````python
auc_simple["Duration"].describe()             # au niveau de chaque formation
summary_years["study_duration"].describe()    # au niveau de chaque individu

summary_years["study_duration"].min()    # formation la plus courte
summary_years["study_duration"].max()    # formation la plus longue
summary_years["study_duration"].mean()   # durée moyenne de formation
summary_years["study_duration"].std()    # écart-type
````

## Barplot

Pour visualiser la distribution de **variables catégorielles** (par exemple, nationalité, discipline, université), on utilise très souvent un **diagramme en barres** :

````python
counts = auc_simple[["Name", "Nationality"]].drop_duplicates()["Nationality"].value_counts().reset_index()
counts.columns = ["Nationality", "n"]


fig, ax = plt.subplots(figsize=(7, 5))
sns.barplot(data=counts, x="Nationality", y="n", order=counts["Nationality"], hue="Nationality", ax=ax,
    palette = {
        "Chinese": sns.color_palette()[3],   # red
        "Japanese": sns.color_palette()[2],  # green
        "Western": sns.color_palette()[0]    # blue
    },
    legend=False)

ax.set_title("Number of Students by Nationality")
ax.set_xlabel(None)
ax.set_ylabel("Count")
fig.text(0.1, -0.02, "D'après 'American University Men of China' (1936)", fontsize=8, style="italic")

plt.show()
````

**Explications**

- `auc_simple[["Name", "Nationality"]].drop_duplicates()` : élimine les doublons — si un même étudiant apparaît plusieurs fois, il ne sera compté qu'une seule fois par nationalité.
- `.value_counts()` calcule le nombre d'observations dans chaque groupe.
- `sns.barplot(x=..., y=..., hue=...)` : `hue` colore les barres selon la nationalité, pour mieux distinguer visuellement les catégories.
- `fig.text(...)` permet d'ajouter une légende de bas de page (source).

Autre exemple avec les disciplines :

````python
fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=counts_field,
    x="Field",
    y="n",
    order=counts_field["Field"],
    hue="Field",
    ax=ax,
    legend=True
)

ax.set_title("Distribution by Field of Study")
ax.set_ylabel("Count")
ax.set_xticks([]) # on enlève les graduations de l'axe horizontal pour ne pas surcharger le graphique
ax.set_xlabel(None) # on enlève le titre de l'axe horizontal pour ne pas surcharger le graphique

plt.show()
````

````python
# Alternative : on peut aussi indiquer le nom des champs comme ceci :

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=counts_field,
    x="Field",
    y="n",
    order=counts_field["Field"],
    hue="Field",
    ax=ax,
    legend=False
)

ax.set_title("Distribution by Field of Study")
ax.set_ylabel("Count")
ax.set_xlabel(None)

ax.tick_params(axis="x", labelrotation=45)  # Rotate field names

plt.show()
````

Autre exemple avec les universités, en inversant les axes (avec `barh`, ou l'argument `orient` en seaborn) :

````python
counts_univ = auc_simple["University"].value_counts().reset_index()
counts_univ.columns = ["University", "n"]
counts_univ = counts_univ[counts_univ["n"] > 5].sort_values("n")

fig, ax = plt.subplots(figsize=(8, 6))
ax.barh(counts_univ["University"], counts_univ["n"])

ax.set_title("Distribution by University — Most Popular Universities (n > 5)")
ax.set_xlabel("Number of curricula")

plt.show()
````

Pour comparer la distribution d'une variable entre deux groupes, on utilise des **facettes**, avec `sns.catplot(..., col=...)`. Par exemple, comparons la distribution des disciplines chez les étudiants chinois et occidentaux :

````python
subset = auc_simple[(auc_simple["Nationality"] != "Japanese") & (auc_simple["Field"].notna())]
grouped = subset.groupby(["Nationality", "Field"]).size().reset_index(name="n")

g = sns.catplot(
    data=grouped, x="Field", y="n", col="Nationality",
    kind="bar", hue="Field", legend=True,
    height=5, aspect=1, sharex=False
)

g.set_xticklabels([])
g.set(xlabel=None, ylabel="Number of curricula")
g.fig.suptitle("Disciplines by Nationality", y=1.05)

plt.show()
````

**Explications**

- `subset = auc_simple[...]` : on exclut les étudiants japonais et les disciplines manquantes, comme précédemment avec `.filter()`.
- `.groupby(["Nationality", "Field"]).size()` : on crée un tableau croisé — on calcule combien d'étudiants appartiennent à chaque couple (nationalité, discipline).
- `sns.catplot(..., col="Nationality")` : on divise le graphique en plusieurs "panneaux", un par nationalité — équivalent de `facet_wrap(~ Nationality)`.

## Scatterplot

Pour croiser **deux variables quantitatives**, on utilise couramment un **nuage de points**. Par exemple, croisons la période de formation et la durée totale des études :

````python
fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(data=summary_years, x="study_start_year", y="study_duration", ax=ax)

ax.set_title("Period and Duration of Study")
ax.set_xlabel("Starting Year")
ax.set_ylabel("Duration (years)")
fig.text(0.1, -0.02, "Based on 'American University Men of China' (1936)", fontsize=8, style="italic")

plt.show()
````

Améliorons un peu l'esthétique de ce graphique. L'argument **`alpha`** permet d'augmenter la transparence pour mieux identifier les zones denses, et l'on peut ajouter une **courbe de tendance** avec `sns.regplot()` ou `lowess=True` :

````python
fig, ax = plt.subplots(figsize=(8, 5))

sns.regplot(
    data=summary_years, x="study_start_year", y="study_duration",
    scatter_kws={"alpha": 0.5}, lowess=True, line_kws={"linewidth": 1.5}, ax=ax
)

ax.set_title("Period and Duration of Study")
ax.set_xlabel("Starting Year")
ax.set_ylabel("Duration (years)")

plt.show()
````

**Explications**

- `sns.regplot(..., lowess=True)` : ajoute une courbe de tendance lissée (méthode LOWESS).
- `scatter_kws={"alpha": 0.5}` : contrôle la transparence des points.

Enfin, on peut modifier la couleur des points, par exemple, **colorier les points par nationalité** pour mieux les distinguer :

````python
fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=summary_years, x="study_start_year", y="study_duration",
    hue="Nationality", alpha=0.7, ax=ax
)

ax.set_title("Period and Duration of Study")
ax.set_xlabel("Starting Year")
ax.set_ylabel("Duration (years)")

plt.show()
````

2. [Python Software Foundation](https://www.python.org/) : site officiel du langage Python.
3. [Pandas — Getting started](https://pandas.pydata.org/docs/getting_started/index.html) : tutoriel officiel de pandas.
4. [Python for Data Analysis (Wes McKinney)](https://wesmckinney.com/book/) : l'ouvrage de référence, par le créateur de pandas, en accès libre.

## Pour aller plus loin

5. [The Programming Historian (en français et en anglais)](https://programminghistorian.org/) : tutoriels de programmation pensés spécifiquement pour les sciences humaines.
6. [Python Graph Gallery](https://python-graph-gallery.com/) : galerie de visualisations avec exemples de code.
7. [Real Python](https://realpython.com/) : tutoriels Python de qualité, tous niveaux.
