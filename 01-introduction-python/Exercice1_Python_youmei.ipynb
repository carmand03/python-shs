## 3. Rationaliser

Le jeu de données contient beaucoup de colonnes (39 au total), dont plusieurs sont redondantes (par exemple, le nom transcrit selon différents systèmes). Simplifions-le pour ne conserver que l'information la plus utile pour la suite.

**Question 7.** Créez un nouveau dataframe `youmei_simple`, en conservant **uniquement** les colonnes suivantes, **dans l'ordre indiqué**, et en les renommant comme proposé dans le tableau ci-dessous. Veillez à ne pas écraser le jeu de données d'origine (`youmei`).

**Question 13.** Combien y a-t-il d'étudiants boursiers par rapport aux étudiants autofinancés ? (Créez une nouvelle variable pour réduire la variable `financement` à deux catégories.)

<div class="alert alert-success" role="alert" style="background-color:#d4edda;padding:10px;border-radius:5px;">
💡 Commencez par afficher les valeurs uniques de <code>financement</code> avec <code>.value_counts()</code> pour voir quelles catégories existent réellement, avant de décider comment les regrouper. Une fonction personnalisée, appliquée avec <code>.apply()</code>, est un bon outil ici (voir section 1.8 du cours).

````
# ✏️ Votre code ici


````

<details>
<summary>▶️ Voir la solution</summary>

```python
youmei_simple["financement"].value_counts()
# Government support, Private support, Partial support, Unknown

def simplifier_financement(f):
    if f == "Government support":
        return "Boursier"
    elif f in ["Private support", "Partial support"]:
        return "Autofinancé"
    else:
        return "Inconnu"

youmei_simple["financement_simple"] = youmei_simple["financement"].apply(simplifier_financement)
youmei_simple["financement_simple"].value_counts()
```

D'autres regroupements sont tout à fait défendables (par exemple, exclure les `Partial support`, ambigus, plutôt que de les rattacher aux autofinancés) — l'essentiel est de **justifier** son choix, comme pour toute catégorisation en sciences sociales.

<div class="alert alert-success" role="alert" style="background-color:#d4edda;padding:10px;border-radius:5px;">
💡 <b>Indices</b>

- Pour (18) et (19) : il s'agit de croiser deux variables catégorielles. Un tableau croisé (<code>pd.crosstab()</code>, éventuellement en pourcentage avec <code>normalize=</code>) et un barplot avec l'argument <code>hue</code> permettent déjà de faire émerger des tendances, sans qu'il soit nécessaire de recourir à des méthodes statistiques plus avancées (nous verrons l'analyse des correspondances, mieux adaptée à ce type de question, lors d'une prochaine séance).
- Pour (20) : un boxplot croisant la discipline (catégorielle) et la durée (quantitative) est l'outil idéal, comme vu en section 5 du cours.
- Pour (21) : un nuage de points (scatterplot) croisant l'année d'arrivée et la durée des études, éventuellement avec une courbe de tendance (<code>sns.regplot(..., lowess=True)</code>), permet de visualiser une évolution au fil du temps.
</div>


````
# ✏️ Votre code ici — utilisez autant de cellules que nécessaire



````

<details>
<summary>▶️ Voir une piste de solution</summary>

```python
# Question 18 : origine géographique et bourse
pd.crosstab(youmei_simple["province_naissance"], youmei_simple["financement_simple"])
# Se concentrer sur les provinces les plus représentées pour un résultat plus lisible :
top_provinces = youmei_simple["province_naissance"].value_counts().head(6).index
subset = youmei_simple[youmei_simple["province_naissance"].isin(top_provinces)]
pd.crosstab(subset["province_naissance"], subset["financement_simple"], normalize="index").round(3) * 100

# Question 19 : financement et discipline
pd.crosstab(youmei_simple["financement_simple"], youmei_simple["discipline"], normalize="index").round(3) * 100

fig, ax = plt.subplots(figsize=(9, 5))
sns.countplot(data=youmei_simple, x="discipline", hue="financement_simple", ax=ax)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right")
plt.show()

# Question 20 : durée des études selon la discipline
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=youmei_simple, x="discipline", y="duree", hue="discipline", legend=False, ax=ax)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right")
ax.set_title("Durée des études selon la discipline")
plt.show()

# Question 21 : durée des études au fil des générations
fig, ax = plt.subplots(figsize=(8, 5))
sns.regplot(data=youmei_simple, x="annee_arrivee_USA", y="duree", lowess=True,
            scatter_kws={"alpha": 0.5}, ax=ax)
ax.set_title("Durée des études selon la génération")
ax.set_xlabel("Année d'arrivée aux États-Unis")
ax.set_ylabel("Durée des études (années)")
plt.show()
```
